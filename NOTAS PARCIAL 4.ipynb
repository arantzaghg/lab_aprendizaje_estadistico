# Shap Values

- Para los shap values se debe de realizar todas las permutaciones posibles para todas las variables, para así calcular la contribución marginal de las variables y sacar el promedio de esto para sacar el shap value. 

## Contribución marginal

- La contribución marginal de cada variable se refiere a cuanto cambia la predicción del modelo al momento de agregar esta misma variable al conjunto de variables ya establecido.

## Pasos

- Primero se debe de considerar una permutación como la siguiente: $[x_2, x_1, x_3]$. Después, se parte de un modelo sin variables $f(∅)$. Se van agregando uno por uno las distintas variables de la siguiente manera: $x_2 = f({x2}) - f(∅)$ (esta es la contribución de $x_2$). Luego se agrega $x_1$ y se calula su contribución: $x1 = f({x_2, x_1}) - f({x_2})$. Lo mismo con $x_3$ $x_3 = f({x_2, x_1, x_3}) - f({x_2, x_1})$. Se repite con todas las permutaciones y se saca el promedio de todas las contribuciones marginales de cada variable y se obtiene el shap value de cada variable de la siguiente manera:

$$
\phi_i = \sum_{S \subseteq N \setminus \{i\}} \frac{|S|! \cdot (n - |S| - 1)!}{n!} \cdot \left[ f(S \cup \{i\}) - f(S) \right]
$$

Finalmente, se calcula la predicción de un modelo de la siguiente manera: $f(x_1, x_2, x_3) = f(∅) + \text{SHAP} x_1 + \text{SHAP} x_2 + \text{SHAP} x_3$


# KMEANS

- Es un algoritmo que sirve para dividir un conjunto de datos en clusters, tomando en cuenta que los elementos dentro de cada gruo tengan gran similitudes entre ellos y que sean diferentes a los otros grupos. Los clusters se forman considerando la cercanía o la distancia que existe entre los puntos a los centros (centroides)

## Pasos

- Tomando en cuenta un plano con muchos puntos disperos, se busca agrupar los datos por cercanía. Primero, se inicia con centros de grupos (K centroides) de manera aleatoria. Se calcula la distancia euclidiana al cuadrado de la siguiente manera:

$$
\|x_i - \mu_k\|^2 = \sum_{j=1}^n (x_{ij} - \mu_{kj})^2
$$

Se le asigna cada punto ($x_1$) al centro ($u_k$) más cercano de la siguiente manera: 

$$ c_i = \arg\min_k \|x_i - \mu_k\|^2 $$

y se recalcula y actualiza cada centro como el promedio de los puntos asignados a ese grupo (esto se hace hasta converger):

$$ \mu_k = \frac{1}{|C_k|} \sum_{x_i \in C_k} x_i $$

Se busca minimizar la suma de las distancias cuadradas dentro de cada cluster, esto se hace de la siguiente manera:

$$ J = \sum_{k=1}^K \sum_{x_i \in C_k} \|x_i - \mu_k\|^2 $$

# PCA

- Es una técnica que busca reducir la dimensionalidad de un conjunto de datos, buscando mantener la mayor cantidad de la varianza original. Esto se hace para no perder información fundamental para la predicción del modelo. 

## Pasos

- Primero se debe de considerar una proyección de una componente:

$$
\text{Proy}_{u_1}(x_i) = u_1^T x_i
$$

Se debe de calcular la varianza que explique la mayor cantidad de varianza posible, para no perder información clave, esto se hace de la siguiente manera:

$$
\text{Varianza} = \frac{1}{N} \sum_{n=1}^N \left( u_1^T x_n - u_1^T \bar{x} \right)^2
$$

Se reorganiza la información, aplican propiedades transpuestas y agrupan términos para simplificar esto de la siguiente manera:

$$
= u_1^T S u_1
$$

En este caso, la S es la matriz de covarianza de los datos y se considera que $u_1$ es la dirección del primer componente principal

Luego, se debe de maximizar la varianza de las proyecciones, para esto se necesita encontrar el vector $u_1$:

$$
\max_{u_1} u_1^T S u_1
$$

Para esto, se aplican los multiplicadores de Lagrange. La función Lagrangiana es la siguiente:

$$
\mathcal{L}(u_1, \lambda) = u_1^T S u_1 - \lambda (u_1^T u_1 - 1)
$$

Se deriva con respecto a $u_1$ y se iguala a cero para maximizar, simplificado se obtiene lo siguiente:

$$
S u_1 = \lambda u_1
$$

Dicho esto, el primer componente principal $u_1$ es el eigenvector con eigenvalor ($\lambda$) más grande, lo cual permite que los componentes expliquen la mayor parte de la varianza posible y simplificar el uso de información. Esto se aplica para cada componente, o sea el segundo eigenvalor más grande es el que resuelve la segunda componente principal.

## Tratamiento continuo en S Learner

- Primero se debe de definir que variable continua se va a utilizar para analizar su efecto. Se debe de entrenar el modelo sobre los datos y a la variable continua elegida se le dan distintos valores. El modelo, de manera iterativa, hace predicciones con los valores. Así, se ve el efecto que tiene esta variable continua sobre la predicción

## S Learner (discreto)

- Lo datos originales se modifican de tal manera que un grupo de estos tendrá el tratamiento 1 y el otro no tendrá tratamiento, es decir tratamiento (0). Se realizan las predicciones con ambos grupos y se restan estas predicciones para obtener el CATE de tener el tratamiento.

## CATE 

- Es el efecto causal del tratamiento para las personas con características específicas

## T Learner

- Los datos originales se modifican de tal manera que se obtienen dos datasets, uno con puros tratamiento 1 y el otro con puros tratamiento 0. Con estos, se realizan dos predicciones separadas con los datos test originales. Se restan las predicciones y se obtiene el CATE del tratamiento. Es importante mencionar que la predicción que se hizo con el dataset de puros tratamiento 1 tendrá una mayor carga hacia el tratamiento, mientras que el otro dataset tendrá mayor carga hacia tratamiento nulo o tratamiento 0.

## X Learner

- Para empezar, se tiene que modificar el dataset original en dos, uno contratamiento 0 y el otro con tratamiento 1. Con estos, se calcula D0 y D1 que son los efectos causales del contrafactual para cada grupo. Un ejemplo es calcular D0, donde se calcula que hubiera pasado si el individuo con tratamiento nunca se le hubiera tratado y viceversa para D1. Después de esto, se entran otros dos modelos para predecir el CATE de ambos grupos mencionados anteriormente. Para esto, entrenas un modelo con los datos con tratamiento como $x$ y D0 con la variable $y$, mientras que el segundo se hace igual pero con sus datos correspondientes. Tras esto, se calcula el propensity score, el cual explica la probabilidad de tener el tratamiento condicional a $x$. Para esto, se hace un modelo de clasificación donde $x$ son los datos originales y $y$ si tienen tratamiento o no. Por último, se usan los modelos que se utilizaron para predecir el CATE y el propensity score para ponderar el CATE dependiendo de la probabilidad de tener o no el tratamiento.
