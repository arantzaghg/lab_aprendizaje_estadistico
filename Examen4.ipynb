## Sección 1: Interpretabilidad con SHAP (25 puntos)

**1.1** (Teoría, 10 pts)  
¿Qué representa un valor SHAP en el contexto de un modelo de machine learning? ¿Cuál es su fundamento teórico?

- Para los shap values se debe de realizar todas las permutaciones posibles para todas las variables, para así calcular la contribución marginal de las variables y sacar el promedio de esto para sacar el shap value. La contribución marginal de cada variable se refiere a cuanto cambia la predicción del modelo al momento de agregar esta misma variable al conjunto de variables ya establecido.Es decir, los Shap values representan cuanto aporta una variable a la predicción del modelo establecido. Su fundamento teórico es el de los juegos cooperativos, es decir, los valores de Shapley, esto se explica a continuación. 

- Primero se debe de considerar una permutación como la siguiente: $[x_2, x_1, x_3]$. Después, se parte de un modelo sin variables $f(∅)$. Se van agregando uno por uno las distintas variables de la siguiente manera: $x_2 = f({x2}) - f(∅)$ (esta es la contribución de $x_2$). Luego se agrega $x_1$ y se calula su contribución: $x1 = f({x_2, x_1}) - f({x_2})$. Lo mismo con $x_3$ $x_3 = f({x_2, x_1, x_3}) - f({x_2, x_1})$. Se repite con todas las permutaciones y se saca el promedio de todas las contribuciones marginales de cada variable y se obtiene el shap value de cada variable de la siguiente manera:

$$
\phi_i = \sum_{S \subseteq N \setminus \{i\}} \frac{|S|! \cdot (n - |S| - 1)!}{n!} \cdot \left[ f(S \cup \{i\}) - f(S) \right]
$$

Finalmente, se calcula la predicción de un modelo de la siguiente manera: $f(x_1, x_2, x_3) = f(∅) + \text{SHAP} x_1 + \text{SHAP} x_2 + \text{SHAP} x_3$

## Sección 2: K-Means Clustering (20 puntos)

**2.1** (Teoría, 10 pts)  
Explica que hace KMeans, el algoritmo, como encuentra clusters, etc.

- KMeans es un algoritmo que sirve para dividir un conjunto de datos en clusters, tomando en cuenta que los elementos dentro de cada grupo tengan alta similitud entre ellos y que sean distintos de los elementos de los otros grupos. Los clusters se forman considerando la cercanía o la distancia que existe entre los puntos a los centros (centroides)

- Tomando en cuenta un plano con muchos puntos disperos, se busca agrupar los datos por cercanía. Primero, se inicia con centros de grupos (K centroides) de manera aleatoria. Se calcula la distancia euclidiana al cuadrado de la siguiente manera:

$$
\|x_i - \mu_k\|^2 = \sum_{j=1}^n (x_{ij} - \mu_{kj})^2
$$

Se le asigna cada punto ($x_1$) al centro ($u_k$) más cercano de la siguiente manera: 

$$ c_i = \arg\min_k \|x_i - \mu_k\|^2 $$

y se recalcula y actualiza cada centro como el promedio de los puntos asignados a ese grupo (esto se hace hasta converger):

$$ \mu_k = \frac{1}{|C_k|} \sum_{x_i \in C_k} x_i $$

Se busca minimizar la suma de las distancias cuadradas dentro de cada cluster, esto se hace de la siguiente manera:

$$ J = \sum_{k=1}^K \sum_{x_i \in C_k} \|x_i - \mu_k\|^2 $$

## Sección 3: PCA – Análisis de Componentes Principales (20 puntos)

**3.1** (Teoría, 5 pts)  
¿Qué significa que la primera componente principal maximiza la varianza?

- La primera componente principal, la cual es $u_1$ es el eigenvector con eigenvalor ($\lambda$) más grande, lo cual permite que los componentes expliquen la mayor parte de la varianza posible y simplificar el uso de información. Dicho esto, la primera componente principal maxmiza la varianza para poder reducir la dimensionalidad del conjunto de datos, esto se hace a través de multiplicadores de Lagrange. Además, el primer componente principal es aquel que explica mayor parte de la varianza, por ende, se esta maximizando el valor de este. Se tiene que la varianza es:

$$
\text{Varianza} = \frac{1}{N} \sum_{n=1}^N \left( u_1^T x_n - u_1^T \bar{x} \right)^2
$$

$$
= u_1^T S u_1
$$

Luego, se debe de maximizar la varianza de las proyecciones:

$$
\max_{u_1} u_1^T S u_1
$$

La función Lagrangiana es la siguiente:

$$
\mathcal{L}(u_1, \lambda) = u_1^T S u_1 - \lambda (u_1^T u_1 - 1)
$$

$$
S u_1 = \lambda u_1
$$

Con esto se puede ver como se necesita maximizar la varianza para encontrar el eigenvalor y eigenvector de la primera componente principal. 


**3.2** (Cálculo, 10 pts)  
Si tienes 10 variables correlacionadas y aplicas PCA, ¿cuántas componentes necesitas para explicar al menos el 90% de la varianza? Describe cómo se respondería esta pregunta 

- La respuesta directa a esta pregunta es, ahorita no se puede saber cuantas componentes son necesarias para explicar al menos el 90% de la varianza. Esto es debido a que depende de cada conjunto de datos. Para obtener los resultados se necesita hacer lo siguiente:

- Primero se debe de considerar una proyección de una componente:

$$
\text{Proy}_{u_1}(x_i) = u_1^T x_i
$$

Se debe de calcular la varianza que explique la mayor cantidad de varianza posible, para no perder información clave, esto se hace de la siguiente manera:

$$
\text{Varianza} = \frac{1}{N} \sum_{n=1}^N \left( u_1^T x_n - u_1^T \bar{x} \right)^2
$$

Se reorganiza la información, aplican propiedades transpuestas y agrupan términos para simplificar esto de la siguiente manera:

$$
= u_1^T S u_1
$$

En este caso, la S es la matriz de covarianza de los datos y se considera que $u_1$ es la dirección del primer componente principal

Luego, se debe de maximizar la varianza de las proyecciones, para esto se necesita encontrar el vector $u_1$:

$$
\max_{u_1} u_1^T S u_1
$$

Para esto, se aplican los multiplicadores de Lagrange. La función Lagrangiana es la siguiente:

$$
\mathcal{L}(u_1, \lambda) = u_1^T S u_1 - \lambda (u_1^T u_1 - 1)
$$

Se deriva con respecto a $u_1$ y se iguala a cero para maximizar, simplificado se obtiene lo siguiente:

$$
S u_1 = \lambda u_1
$$

Dicho esto, se tendría que ver cuanto explica cada componente de la varianza, y así se podría saber la cantidad de componentes necesarios para explicar al menos el 90% de la varianza. 


**3.3** (Aplicación, 5 pts)  
¿Tiene sentido usar PCA y usar todas las componentes para predecir? Justifica tu respuesta.

- No tiene sentido usar PCA y usar todas las componentes para predecir. El objetivo del PCA es reducir la dimensionalidad de un conjunto de datos, buscando mantener la mayor cantidad de la varianza original. Esto se hace para no perder información fundamental para la predicción del modelo. Si se usaran todos los componentes, no se esta optimizando el uso de información y datos, ni se estaría reduciendo la dimensionalidad. Se deben de usar los componentes más importantes y que expliquen la mayor parte de la varianza posible, sin necesidad de utilizar todos los componentes.

## Sección 4: Causalidad y Meta-Learners (35 puntos)

**4.1** (Conceptual, 10 pts)  
Define el estimando CATE y da un ejemplo de aplicación en negocios.

- El CATE es el efecto causal promedio del tratamiento para las personas con características específicas. En otras palabras, nos permite ver el efecto que tiene un tratamiento sobre un grupo de personas en específico. Esto se aplica directamente en casos de AB testing. Un ejemplo de aplicación en negocios sería probar un aumento de precios de un producto de una empresa. Tomando en cuenta que al grupo A se le da el precio original y al grupo B el precio aumentado, el CATE nos ayudaría a analizar como este cambio en el precio afecta cada grupo de personas. Por ejemplo, el CATE nos puede indicar que el efecto causal promedio de este cambio puede ser malo para personas con cierta sensibilidad al precio o viceversa. 

**4.2** (Comparación, 15 pts)  
Explica las diferencias entre los siguientes enfoques para estimar CATE:
- S-Learner
- T-Learner  
- X-Learner  

Incluye ventajas y desventajas.

- Empezando con S Learner, este solamente entrena a un modelo original y se realizan predicciones considerando que todos tienen 1, es decir que se les aplicó el tratamiento. En este caso, se realizan estas predicciones con ambos grupos y se restan para obtener el CATE de si tener el tratamiento. La ventaja de este método es que es sencillo y más facil de realizar. Además, usa toda la información dada de manera directa. Sin embargo, la desventaja sería que este método puede fallar al momento de capturar el efecto del tratamiento. Esto depende altamente del modelo original y como es que esta la información de esta variable.


- Por otro lado, en el método T Learner los datos originales se modifican de tal manera que se obtienen dos datasets, uno con puros tratamiento 1 y el otro con puros tratamiento 0. Con estos, se realizan dos predicciones separadas con los datos test originales. Se restan las predicciones y se obtiene el CATE del tratamiento. Una ventaja clave de este método es que se modela de manera separada pro cada grupo, esto permite darle mayor peso a las características específicas del grupo (tratamiento 1 o tratamiento 0). Dicho esto, la desventaja sería que la predicción que se hizo con el dataset de puros tratamiento 1 tendrá una mayor carga hacia el tratamiento, mientras que el otro dataset tendrá mayor carga hacia tratamiento nulo o tratamiento 0. Esto podría afectar los resultados y la precisión.


- Por último, en el método X Learner se tiene que modificar el dataset original en dos, uno con tratamiento 0 y el otro con tratamiento 1. Con estos, se calcula D0 y D1 que son los efectos causales del contrafactual para cada grupo. Después de esto, se entran otros dos modelos para predecir el CATE de ambos grupos mencionados anteriormente. Para esto, entrenas un modelo con los datos con tratamiento como $x$ y D0 con la variable $y$, mientras que el segundo se hace igual pero con sus datos correspondientes. Tras esto, se calcula el propensity score, el cual explica la probabilidad de tener el tratamiento condicional a $x$. Para esto, se hace un modelo de clasificación donde $x$ son los datos originales y $y$ si tienen tratamiento o no. Por último, se usan los modelos que se utilizaron para predecir el CATE y el propensity score para ponderar el CATE dependiendo de la probabilidad de tener o no el tratamiento. Dicho esto, la principal ventaja de este método es que su complejidad permite resultados mejores y más precisios. El CATE en este caso sería mucho mejor dado la manera que se realiza el método. La desventaja sería que puede ser un proceso tardado y sería algo facil fallar al utilizarlo.
